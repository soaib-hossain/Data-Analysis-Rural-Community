# Rural Ontario Community Wellbeing Analysis
**Data Source:** Rural Ontario Institute (ROI) Community Wellbeing Dashboard  
**File:** `ROI_All_Data_Sources.xlsx`  

This notebook analyzes wellbeing indicators across 6 categories:
- Demographic, Economic, Environment, Health, Housing, Society

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_rows', 100)
pd.set_option('display.max_colwidth', 80)
sns.set_theme(style='whitegrid')

print('Libraries loaded successfully!')
```

## 1. Load the Data

```python
# Load the Excel file
raw = pd.read_excel('ROI_All_Data_Sources.xlsx', sheet_name='All Sources', header=2)
raw.columns = ['Category', 'Indicator', 'Data_Source']

# Drop fully empty rows and reset index
df = raw.dropna(how='all').reset_index(drop=True)

# Remove the repeated header row
df = df[df['Category'] != 'Category'].reset_index(drop=True)

print(f'Total indicators: {len(df)}')
df.head(10)
```

## 2. Explore the Dataset Structure

```python
# Forward-fill categories (some rows share a category label)
df['Category'] = df['Category'].ffill()

print('Categories and indicator counts:')
print(df['Category'].value_counts())
```

```python
# Data sources used across all categories
print('Unique data sources:')
print(df['Data_Source'].dropna().unique())
```

## 3. Visualize: Number of Indicators per Category

```python
category_counts = df['Category'].value_counts()

plt.figure(figsize=(9, 5))
bars = sns.barplot(x=category_counts.values, y=category_counts.index, palette='viridis')
plt.title('Number of Indicators per Wellbeing Category', fontsize=14)
plt.xlabel('Number of Indicators')
plt.ylabel('Category')

for i, v in enumerate(category_counts.values):
    plt.text(v + 0.1, i, str(v), va='center')

plt.tight_layout()
plt.savefig('indicators_per_category.png', dpi=150)
plt.show()
print('Chart saved as indicators_per_category.png')
```

## 4. Indicators by Category

```python
# View all indicators for a specific category
# Change the category name to explore others:
# 'Demographic', 'Economic', 'Environment', 'Health', 'Housing', 'Society'

selected_category = 'Economic'
subset = df[df['Category'] == selected_category][['Indicator', 'Data_Source']]
print(f'--- {selected_category} Indicators ---')
subset.reset_index(drop=True)
```

## 5. Data Sources Breakdown

```python
source_counts = df['Data_Source'].value_counts()

plt.figure(figsize=(10, 5))
sns.barplot(x=source_counts.values, y=source_counts.index, palette='mako')
plt.title('Indicators by Data Source', fontsize=14)
plt.xlabel('Number of Indicators')
plt.ylabel('Data Source')
plt.tight_layout()
plt.savefig('indicators_by_source.png', dpi=150)
plt.show()
print('Chart saved as indicators_by_source.png')
```

## 6. Full Indicator Reference Table

```python
# Full sorted table of all indicators
df.sort_values(['Category', 'Indicator']).reset_index(drop=True)
```
